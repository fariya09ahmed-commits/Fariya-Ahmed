```python
from sklearn.preprocessing import LabelEncoder
label_encoder=LabelEncoder()
#convert mainroad column
df['mainroad_encoded'] = label_encoder.fit_transform(df['mainroad'])
print("\nAfter Label Encoding 'mainroad':")
print(df[['mainroad', 'mainroad_encoded']].head())
```

```python

X = df.drop('furnishingstatus', axis=1)
y = df['furnishingstatus']

print("X has all housing information (12 columns)")
print("y has furnishingstatus yes/no (1 column)")

# Split: 80% for training, 20% for testing
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.2,
    random_state=20
)

print(f"\nWe will train with: {len(X_train)} housing")
print(f"We will test with: {len(X_test)} housing")
```

```python
from sklearn.preprocessing import StandardScaler

# Columns to scale
cols_to_scale = ['area', 'bedrooms']

# Create scaler
scaler = StandardScaler()

# Scale training data (fit + transform)
X_train_scaled = X_train.copy()
X_train_scaled[cols_to_scale] = scaler.fit_transform(X_train[cols_to_scale])

# Scale test data (just transform)
X_test_scaled = X_test.copy()
X_test_scaled[cols_to_scale] = scaler.transform(X_test[cols_to_scale])

# Check examples
print("\nAfter scaling 'area' and 'bedrooms':")
print(f"First housing area before scaling: {X_train['area'].iloc[0]}")
print(f"First housing area after scaling: {X_train_scaled['area'].iloc[0]:.2f}")
print(f"First bedrooms before scaling: {X_train['bedrooms'].iloc[0]}")
print(f"First bedrooms after scaling: {X_train_scaled['bedrooms'].iloc[0]:.2f}")
```

```python
from sklearn.linear_model import LogisticRegression

model = LogisticRegression(max_iter=100)
model.fit(X_train_scaled, y_train)
print("✅ Model trained!")

train_score = model.score(X_train_scaled, y_train)
print(f"Training accuracy: {train_score:.1%}")

test_score = model.score(X_test_scaled, y_test)
print(f"Test accuracy: {test_score:.1%}")
```
