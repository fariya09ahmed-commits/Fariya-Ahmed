```python
#assignment-02

# Parameters
n = 6
L = 5.0
K = 2.0
S = 3.0
TA = 5.0
TB = 15.0
niter = 3

dx = L / (n - 1)

x = np.array([i * dx for i in range(n)])
AA = -S / (2 * K)
c2 = TA
c1 = (TB - TA - AA * L**2) / L
T = AA * x**2 + c1 * x + c2

print("Iteration    Error")

for iter in range(1, niter + 1):
    Told = T.copy()

    a = np.ones(n-2)
    b = -2.0 * np.ones(n-2)
    c = np.ones(n-2)
    d = -(S / K) * dx**2 * np.ones(n-2)

    d[0] -= T[0]
    d[-1] -= T[-1]

    cp = np.zeros(n-2)
    dp = np.zeros(n-2)

    cp[0] = c[0] / b[0]
    dp[0] = d[0] / b[0]

    for i in range(1, n-2):
        denom = b[i] - a[i] * cp[i-1]
        cp[i] = c[i] / denom
        dp[i] = (d[i] - a[i] * dp[i-1]) / denom

    T[n-2] = dp[-1]
    for i in range(n-4, -1, -1):
        T[i+1] = dp[i] - cp[i] * T[i+2]

    T[0] = TA
    T[-1] = TB

    err = np.sum(np.abs(T - Told))
    print(f"{iter:5d}    {err:12.6e}")

print("\n*** TEMPERATURE DISTRIBUTION ***\n")
print("   x     Temperature")
for i in range(n):
    print(f"{x[i]:5.2f}   {T[i]:10.4f}")
```
