## Understand the Data

```python
# Basic exploration
print("Dataset Shape:", df.shape)
print("\nFirst 5 rows:")
print(df.head())

print("\nDataset Info:")
print(df.info())

print("\nBasic Statistics:")
print(df.describe())

print("\nCheck for missing values:")
print(df.isnull().sum())

print("\nTarget variable distribution:")
print(df['Outcome'].value_counts())

```

## Line Graph

```python
# First 10 math scores
scores = df['Age'].head(10)

# Create x-axis as 1 to 10
x = range(1, 11)   # or list(range(1, 11))

plt.plot(x, scores, marker='o')
plt.title('Age of First 10 ')
plt.xlabel('Age')
plt.ylabel('Outcome')
#plt.xticks(range(1, 11))
plt.show()
```

# Bar Chart

```python
# Count Area values
gender_count = df['Area'].value_counts()

# Simple bar chart
plt.figure(figsize=(12, 6)) # Adjust figure size for better readability with many bars
plt.bar(gender_count.index, gender_count.values)
plt.title('Count of Each Area')
plt.xlabel('Area')
plt.ylabel('Count')
plt.xticks(rotation=90) # Rotate x-axis labels for better readability
plt.tight_layout() # Adjust layout to prevent labels from overlapping
plt.show()
```

## Histogram

```python
# Age distribution
plt.hist(df['Age'])
plt.title('Age Distribution')
plt.show()
```

```python
# With more bins
plt.hist(df['Age'], bins=20, color='green')
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.title('Age Distribution (with more bins)')
plt.show()
```

```python
# Test preparation
test_prep = df['Gender'].value_counts()

plt.pie(test_prep.values, labels=['Gender','AreaType'])
plt.title('Area in Gender')
# With percentages
plt.pie(test_prep.values, labels=['Gender', 'AreaType'], autopct='%1.1f%%')
plt.title('Area in Gender')
plt.show()
```

##  Multiple plot

```python
# Create 2x2 grid
fig, axes = plt.subplots(2, 2, figsize=(10, 8))

# Plot 1: Histogram
axes[0, 0].hist(df['Age'], bins=15)
axes[0, 0].set_title('Age distribution')

# Plot 2: Bar chart
df['AreaType'].value_counts().plot(kind='bar', ax=axes[0, 1])
axes[0, 1].set_title('AreaType')

# Plot 3: Scatter
axes[1, 0].scatter(df['NS1'][:50], df['IgM'][:50])
axes[1, 0].set_title('NS1 VS IgM')

# Plot 4: Line
axes[1, 1].plot(df['Age'][:30])
axes[1, 1].set_title('Age ')

plt.tight_layout()
plt.show()
```

## Seaborn

```python
sns.histplot(df['AreaType'])
plt.title('AreaType with Seaborn')
plt.show()

# With KDE curve
sns.histplot(df['Outcome'], kde=True)
plt.show()
```

## Count Plot

```python
# Count by District
sns.countplot(data=df, x='Gender')
plt.grid(True, which="both", linestyle="--", linewidth=0.7, alpha=0.7)
plt.title('District Count')
plt.xticks(rotation=45, ha='right') # Rotate labels for better readability
plt.tight_layout() # Adjust layout to prevent labels from overlapping
plt.show()
```

```python
# By parent education
sns.barplot(data=df, x='Age', y='Outcome')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

## Bar plot with average

```python
# Math score by gender
sns.kdeplot(data=df, x='Age', hue='Outcome')
plt.grid(True, which="both", linestyle="--", linewidth=0.7, alpha=0.7)
plt.title('Age by Outcome')
plt.show()
```

## Heatmap

```python
# Correlation between scores
scores = df[['NS1', 'IgG', 'IgM']]
correlation = scores.corr()
print(correlation)
```

```python
# With numbers
sns.heatmap(correlation, annot=True)
plt.title('Correlations')
plt.show()
```
