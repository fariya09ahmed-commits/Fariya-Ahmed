## **conditional**

```python
#assg-1
age=int(input('Enter your age'))
if age >= 18:
  print("you are an old")
else:
  print("you are not an old")
```

```python
#assg-2
number=int(input('Enter your number'))
if number >= 0:
  print("positive number")
elif number <= 0:
  print("negative number")
else:
  print("the number is zero")
```

```python
#assg-3
number=int(input('Enter the number'))
if number % 2 == 0:
  print("even number")
else:
  print("odd number")
```

```python
#assg-4
age=int(input('Enter the user age'))
if age <= 0:
  print("age not found")
elif age <= 12:
  print("child")
elif age <= 19:
  print("teenager")
elif age <= 64:
  print("adult")
else:
  print("senior")
```

## loop and function

```python
#assg-1
def sum_of_natural_number(n):
    total = 0
    for i in range(1, n+1):
        total= total+i
    return total
result = sum_of_natural_number(5)
print(f"the sum is {result}")
```

```python
num=int(input('Enter any number'))
fact=1
for i in range(1,num+1):
 fact=fact*i
print("Factorial of",num,"is",fact)

```

```python
for i in range(1, 10):
  for j in range(1, 10):
    product = i * j
    print(f"{i} × {j} = {product}")
  print("-----")
```
