```python
file_path = ('/content/drive/MyDrive/gym_membership.csv')
gym = pd.read_csv(file_path)
```

```python
#assg-1
abonoment_count=gym['abonoment_type'].value_counts()
plt.bar(['premium','standard'],abonoment_count.values)
plt.title('Gym Membership Types')
plt.show()
```

```python
#assg-2
plt.hist(gym['avg_time_in_gym'],color='green')
plt.title('Avg Time in Gym')
plt.show()
```

```python
#assg-3
plt.scatter(gym['Age'],gym['visit_per_week'])
plt.grid(True,which="both",linestyle="--",linewidth=0.7,alpha=0.7)
plt.xlabel('Age')
plt.ylabel('Visit_per_week')
plt.title('Age vs Visit Per Week')
plt.show()
```

```python
#assg-1
sns.boxplot(data=gym,x='gender',y='Age')
plt.grid(True,which="both",linestyle="--",linewidth=0.7,alpha=0.7)
plt.title('Gender vs Age')
plt.show()
```

```python
#assg-2
sns.countplot(data=gym,x='days_per_week')
plt.grid(True,which="both",linestyle="--",linewidth=0.7,alpha=0.7)
plt.title('Days Per Week')
plt.show()
```

```python
#assg-3
sns.barplot(data=gym,x='abonoment_type',y='visit_per_week')
plt.grid(True,which="both",linestyle="--",linewidth=0.7,alpha=0.7)
plt.title('Average Visits by Membership Type')
plt.show()
```
