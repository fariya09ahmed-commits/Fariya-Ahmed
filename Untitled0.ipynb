```python
# STEP 3 — IQR-Based Outlier Removal (§3.2.2)
# ══════════════════════════════════════════════════════════════
def iqr_clean(X, y, cols, factor=1.5):
    mask = pd.Series(True, index=X.index)
    for c in cols:
        Q1, Q3 = X[c].quantile(0.25), X[c].quantile(0.75)
        IQR = Q3 - Q1
        mask &= X[c].between(Q1 - factor * IQR, Q3 + factor * IQR)
    return X[mask].copy(), y[mask].copy()

outlier_cols = ['Lymphocytes(%)', 'Neutrophils(%)', 'RBC',
                'HCT(%)', 'Total Platelet Count(/cumm)',
                'Total WBC count(/cumm)']
outlier_cols = [c for c in outlier_cols if c in X_all.columns]

X_clean, y_clean = iqr_clean(X_all, y_all, outlier_cols)
print(f"Before: {len(X_all)}  →  After: {len(X_clean)}  "
      f"(removed {len(X_all) - len(X_clean)} outliers)")
print(f"Class after: Positive={y_clean.sum()}  Negative={(y_clean == 0).sum()}")
```

```python
# Fig. 5 — Data Distribution After Outlier Removal
# ══════════════════════════════════════════════════════════════
fig, axes = plt.subplots(2, 4, figsize=(16, 7))
axes = axes.flatten()

for i, feat in enumerate(num_feats[:8]):
    ax = axes[i]
    try:
        X_clean[y_clean == 1][feat].dropna().plot.kde(
            ax=ax, color='#7B2D8B', linewidth=1.8, label='Positive')
        X_clean[y_clean == 0][feat].dropna().plot.kde(
            ax=ax, color='#FFD700', linewidth=1.8, label='Negative')
    except Exception:
        pass
    Q1, Q3 = X_clean[feat].quantile(0.25), X_clean[feat].quantile(0.75)
    IQR = Q3 - Q1
    ax.axvline(Q1 - 1.5 * IQR, color='green', ls='--', lw=1, alpha=0.6)
    ax.axvline(Q3 + 1.5 * IQR, color='green', ls='--', lw=1, alpha=0.6)
    ax.set_title(f'{feat} (Cleaned)', fontsize=9)
    ax.set_xlabel('')

plt.suptitle('Fig. 5 — Data Distribution After IQR Outlier Removal',
             fontsize=11, y=1.01)
plt.tight_layout()
save('fig05_outliers_after')
plt.show()
```

```python
# STEP 4 — Stratified Train/Test Split & StandardScaler (§3.2.3)
# ══════════════════════════════════════════════════════════════
X_train_r, X_test_r, y_train, y_test = train_test_split(
    X_clean, y_clean, test_size=0.20,
    random_state=SEED, stratify=y_clean)

scaler = StandardScaler()
X_train_s = pd.DataFrame(scaler.fit_transform(X_train_r), columns=FEATURES)
X_test_s  = pd.DataFrame(scaler.transform(X_test_r),      columns=FEATURES)

print(f"Train: {len(X_train_s)}  |  Test: {len(X_test_s)}")
print(f"Train class: Pos={y_train.sum()} Neg={(y_train==0).sum()}")
print(f"Test  class: Pos={y_test.sum()}  Neg={(y_test==0).sum()}")
```

```python
# STEP 5 — SMOTE Resampling (§3.2.4) & Fig. 6
# ══════════════════════════════════════════════════════════════
before_counts = pd.Series(y_train).value_counts().sort_index()

smote = SMOTE(random_state=SEED)
X_res, y_res = smote.fit_resample(X_train_s, y_train)

after_counts = pd.Series(y_res).value_counts().sort_index()

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(10, 4))

ax1.bar(['Negative (0)', 'Positive (1)'],
        [before_counts.get(0, 0), before_counts.get(1, 0)],
        color=['#DD8452', '#4C72B0'], edgecolor='k')
for i, v in enumerate([before_counts.get(0, 0), before_counts.get(1, 0)]):
    ax1.text(i, v + 2, str(v), ha='center', fontweight='bold')
ax1.set_title('(a) Before SMOTE')
ax1.set_ylabel('Count')

ax2.bar(['Negative (0)', 'Positive (1)'],
        [after_counts.get(0, 0), after_counts.get(1, 0)],
        color=['#55A868', '#55A868'], edgecolor='k')
for i, v in enumerate([after_counts.get(0, 0), after_counts.get(1, 0)]):
    ax2.text(i, v + 2, str(v), ha='center', fontweight='bold')
ax2.set_title('(b) After SMOTE')
ax2.set_ylabel('Count')

plt.suptitle('Fig. 6 — Class Distribution Before & After SMOTE',
             fontsize=11)
plt.tight_layout()
save('fig06_smote_distribution')
plt.show()
print(f"Before: {before_counts.to_dict()}  →  After: {after_counts.to_dict()}")
```

```python
# STEP 6 — Model Definitions (§3.3 Proposed DengueStackX)
# ══════════════════════════════════════════════════════════════
base_learners = [
    ('rf',   RandomForestClassifier(n_estimators=500, random_state=SEED,
                                     n_jobs=-1)),
    ('et',   ExtraTreesClassifier(n_estimators=500, random_state=SEED,
                                   n_jobs=-1)),
    ('xgb',  XGBClassifier(n_estimators=300, max_depth=6,
                            learning_rate=0.05, subsample=0.8,
                            colsample_bytree=0.8, random_state=SEED,
                            use_label_encoder=False,
                            eval_metric='logloss', verbosity=0)),
    ('lgbm', LGBMClassifier(n_estimators=300, learning_rate=0.05,
                             num_leaves=31, random_state=SEED, verbose=-1)),
    ('svm',  SVC(kernel='rbf', C=10, gamma='scale',
                  probability=True, random_state=SEED)),
]

meta_learner = LogisticRegression(max_iter=2000, C=5.0, random_state=SEED)

DengueStackX = StackingClassifier(
    estimators      = base_learners,
    final_estimator = meta_learner,
    stack_method    = 'predict_proba',
    cv              = 5,
    n_jobs          = -1,
    passthrough     = True,
)

ALL_MODELS = {
    'Logistic Regression' : LogisticRegression(max_iter=2000,
                                                random_state=SEED),
    'Decision Tree'       : DecisionTreeClassifier(criterion='entropy',
                                                    max_depth=10,
                                                    min_samples_split=5,
                                                    random_state=SEED),
    'Random Forest'       : RandomForestClassifier(n_estimators=500,
                                                    random_state=SEED,
                                                    n_jobs=-1),
    'SVM'                 : SVC(kernel='rbf', C=10, gamma='scale',
                                 probability=True, random_state=SEED),
    'KNN'                 : KNeighborsClassifier(n_neighbors=5,
                                                  weights='distance', p=2),
    'XGBoost'             : XGBClassifier(n_estimators=300, max_depth=6,
                                           learning_rate=0.05,
                                           random_state=SEED,
                                           use_label_encoder=False,
                                           eval_metric='logloss',
                                           verbosity=0),
    'LightGBM'            : LGBMClassifier(n_estimators=300,
                                            learning_rate=0.05,
                                            random_state=SEED, verbose=-1),
    'ExtraTrees'          : ExtraTreesClassifier(n_estimators=500,
                                                  random_state=SEED,
                                                  n_jobs=-1),
    'DengueStackX'        : DengueStackX,
}

print(f"{len(ALL_MODELS)} models defined.")
print(f"Base learners in DengueStackX: {[n for n, _ in base_learners]}")
print(f"Meta-learner: LogisticRegression(C=5.0) with passthrough=True")
```

```python
# Table 3 — Stratified 10-Fold Cross-Validation (SMOTE Data)
# ══════════════════════════════════════════════════════════════
skf = StratifiedKFold(n_splits=10, shuffle=True, random_state=SEED)
cv_rows = []

for name, model in ALL_MODELS.items():
    cv = cross_validate(model, X_res, y_res, cv=skf, n_jobs=-1,
                        scoring=['accuracy','precision','recall','f1','roc_auc'])
    cv_rows.append({
        'Model'     : name,
        'Accuracy'  : f"{cv['test_accuracy'].mean()*100:.2f} ± {cv['test_accuracy'].std()*100:.2f}",
        'Precision' : f"{cv['test_precision'].mean()*100:.2f} ± {cv['test_precision'].std()*100:.2f}",
        'Recall'    : f"{cv['test_recall'].mean()*100:.2f} ± {cv['test_recall'].std()*100:.2f}",
        'F1-Score'  : f"{cv['test_f1'].mean()*100:.2f} ± {cv['test_f1'].std()*100:.2f}",
        'ROC-AUC'   : f"{cv['test_roc_auc'].mean()*100:.2f} ± {cv['test_roc_auc'].std()*100:.2f}",
    })

cv_df = pd.DataFrame(cv_rows)
cv_df.to_csv(f'{OUT}/table03_10fold_cv.csv', index=False)
cv_df
```

```python
# Table 4 — Test-Set Classification Performance
# ══════════════════════════════════════════════════════════════
test_rows = []
fitted_models = {}

for name, model in ALL_MODELS.items():
    model.fit(X_res, y_res)
    fitted_models[name] = model
    yp    = model.predict(X_test_s)
    yprob = model.predict_proba(X_test_s)[:, 1]

    acc   = accuracy_score(y_test, yp)
    prec  = precision_score(y_test, yp, zero_division=0)
    rec   = recall_score(y_test, yp, zero_division=0)
    f1v   = f1_score(y_test, yp, zero_division=0)
    spec  = recall_score(y_test, yp, pos_label=0, zero_division=0)
    auc   = roc_auc_score(y_test, yprob)
    mcc   = matthews_corrcoef(y_test, yp)
    kappa = cohen_kappa_score(y_test, yp)
    mae   = mean_absolute_error(y_test, yp)
    rmse  = np.sqrt(mean_squared_error(y_test, yp))
    loglv = log_loss(y_test, yprob)
    rae   = mae / mean_absolute_error(y_test,
                np.full(len(y_test), y_test.mean()))

    test_rows.append({
        'Model'       : name,
        'Accuracy'    : f"{acc*100:.2f}%",
        'Precision'   : f"{prec*100:.2f}%",
        'Recall'      : f"{rec*100:.2f}%",
        'Specificity' : f"{spec*100:.2f}%",
        'F1-Score'    : f"{f1v*100:.2f}%",
        'ROC-AUC'     : f"{auc:.4f}",
        'MCC'         : f"{mcc:.4f}",
        'Kappa'       : f"{kappa:.4f}",
        'Log Loss'    : f"{loglv:.4f}",
        'MAE'         : f"{mae:.4f}",
        'RMSE'        : f"{rmse:.4f}",
        'RAE'         : f"{rae:.4f}",
    })

test_df = pd.DataFrame(test_rows)
test_df.to_csv(f'{OUT}/table04_test_performance.csv', index=False)
test_df
```

```python
# Table 5 — Performance: With vs Without Outlier Removal
# ══════════════════════════════════════════════════════════════
X_tr_raw, X_ts_raw, y_tr_raw, y_ts_raw = train_test_split(
    X_all, y_all, test_size=0.20, random_state=SEED, stratify=y_all)

scaler_raw = StandardScaler()
X_tr_raw_s = pd.DataFrame(scaler_raw.fit_transform(X_tr_raw), columns=FEATURES)
X_ts_raw_s = pd.DataFrame(scaler_raw.transform(X_ts_raw),     columns=FEATURES)
X_res_raw, y_res_raw = SMOTE(random_state=SEED).fit_resample(
    X_tr_raw_s, y_tr_raw)

cmp_models = {
    'Logistic Regression': LogisticRegression(max_iter=2000, random_state=SEED),
    'Decision Tree': DecisionTreeClassifier(criterion='entropy', max_depth=10,
                                             random_state=SEED),
    'Random Forest': RandomForestClassifier(n_estimators=500, random_state=SEED,
                                             n_jobs=-1),
    'SVM': SVC(kernel='rbf', C=10, gamma='scale', probability=True,
                random_state=SEED),
    'KNN': KNeighborsClassifier(n_neighbors=5, weights='distance', p=2),
    'DengueStackX': StackingClassifier(
        estimators=base_learners,
        final_estimator=LogisticRegression(max_iter=2000, C=5.0,
                                            random_state=SEED),
        stack_method='predict_proba', cv=5, n_jobs=-1, passthrough=True),
}

cmp_rows = []
for name, m in cmp_models.items():
    m.fit(X_res_raw, y_res_raw)
    yp_w  = m.predict(X_ts_raw_s)
    acc_w = accuracy_score(y_ts_raw, yp_w)
    f1_w  = f1_score(y_ts_raw, yp_w, zero_division=0)
    m2 = clone(m)
    m2.fit(X_res, y_res)
    yp_c  = m2.predict(X_test_s)
    acc_c = accuracy_score(y_test, yp_c)
    f1_c  = f1_score(y_test, yp_c, zero_division=0)
    cmp_rows.append({
        'Model': name,
        'Acc (With Outliers)': f"{acc_w*100:.2f}%",
        'Acc (Without)': f"{acc_c*100:.2f}%",
        'Δ Acc': f"{(acc_c - acc_w)*100:+.2f}%",
        'F1 (With Outliers)': f"{f1_w*100:.2f}%",
        'F1 (Without)': f"{f1_c*100:.2f}%",
        'Δ F1': f"{(f1_c - f1_w)*100:+.2f}%",
    })

cmp_df = pd.DataFrame(cmp_rows)
cmp_df.to_csv(f'{OUT}/table05_outlier_comparison.csv', index=False)
cmp_df
```

```python
# ══════════════════════════════════════════════════════════════
# Fig. 10 — SHAP Explainability Analysis (§4.2)
# ══════════════════════════════════════════════════════════════
stack_model = fitted_models['DengueStackX']
X_explain = pd.DataFrame(X_test_s.values, columns=FEATURES)

bg = shap.sample(X_explain, min(100, len(X_explain)), random_state=SEED)
explainer = shap.KernelExplainer(stack_model.predict_proba, bg, link='logit')
shap_vals = explainer.shap_values(X_explain.iloc[:50], nsamples=100)

if isinstance(shap_vals, list):
    sv_pos = shap_vals[1]
elif shap_vals.ndim == 3:
    sv_pos = shap_vals[:, :, 1]
else:
    sv_pos = shap_vals

# (a) Beeswarm
fig, ax = plt.subplots(figsize=(9, 6))
plt.sca(ax)
shap.summary_plot(sv_pos, X_explain.iloc[:50].values,
                  feature_names=FEATURES, show=False,
                  plot_type='dot', max_display=len(FEATURES))
ax.set_title('Fig. 10(a) — SHAP Beeswarm: Global Feature Impact')
plt.tight_layout()
save('fig10a_shap_beeswarm')
plt.show()

# (b) Bar
fig, ax = plt.subplots(figsize=(9, 5))
plt.sca(ax)
shap.summary_plot(sv_pos, X_explain.iloc[:50].values,
                  feature_names=FEATURES, show=False,
                  plot_type='bar', max_display=len(FEATURES))
ax.set_title('Fig. 10(b) — SHAP Bar: Mean |SHAP| Feature Importance')
plt.tight_layout()
save('fig10b_shap_bar')
plt.show()
```

```python
# ══════════════════════════════════════════════════════════════
# Fig. 11 — SHAP Waterfall Plots for Individual Predictions
# ══════════════════════════════════════════════════════════════
pos_idx = np.where(y_test.values == 1)[0][0]
neg_idx = np.where(y_test.values == 0)[0][0]

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

for ax, idx, cls_name in [(ax1, pos_idx, 'Positive'),
                           (ax2, neg_idx, 'Negative')]:
    if isinstance(shap_vals, list):
        sv = shap_vals[1][idx] if cls_name == 'Positive' else shap_vals[0][idx]
    elif shap_vals.ndim == 3:
        sv = shap_vals[idx, :, 1] if cls_name == 'Positive' else shap_vals[idx, :, 0]
    else:
        sv = shap_vals[idx]
    feat_vals = X_explain.iloc[idx]
    colors = ['#d73027' if v > 0 else '#4575b4' for v in sv]
    order  = np.argsort(np.abs(sv))[::-1]
    ys     = np.arange(len(order))
    ax.barh(ys, sv[order], color=np.array(colors)[order],
            edgecolor='k', linewidth=0.4)
    ax.set_yticks(ys)
    ax.set_yticklabels([f"{FEATURES[i]} = {feat_vals.iloc[i]:.2f}"
                         for i in order], fontsize=8)
    ax.axvline(0, color='black', linewidth=0.8)
    ax.set_xlabel('SHAP Value')
    ax.set_title(f'({chr(97 + (0 if cls_name == "Positive" else 1))}) '
                 f'True: {cls_name}')

plt.suptitle('Fig. 11 — SHAP Waterfall: Individual Prediction Explanation',
             fontsize=12)
plt.tight_layout()
save('fig11_shap_waterfall')
plt.show()
```

```python
# ══════════════════════════════════════════════════════════════
# Table 6 — Sensitivity Analysis: Feature Ablation Study
# ══════════════════════════════════════════════════════════════
key_feats = ['Lymphocytes(%)', 'Total Platelet Count(/cumm)',
             'HCT(%)', 'Total WBC count(/cumm)', 'Neutrophils(%)']
key_feats = [f for f in key_feats if f in FEATURES]

sens_rows = []
for feat in key_feats:
    remaining = [f for f in FEATURES if f != feat]
    X_r_drop  = X_res[remaining]
    X_t_drop  = X_test_s[remaining]

    # Rebuild base learners for reduced features
    bl_drop = [
        ('rf',   RandomForestClassifier(n_estimators=500, random_state=SEED, n_jobs=-1)),
        ('et',   ExtraTreesClassifier(n_estimators=500, random_state=SEED, n_jobs=-1)),
        ('xgb',  XGBClassifier(n_estimators=300, max_depth=6, learning_rate=0.05,
                                random_state=SEED, use_label_encoder=False,
                                eval_metric='logloss', verbosity=0)),
        ('lgbm', LGBMClassifier(n_estimators=300, learning_rate=0.05,
                                 random_state=SEED, verbose=-1)),
        ('svm',  SVC(kernel='rbf', C=10, gamma='scale', probability=True,
                      random_state=SEED)),
    ]

    test_models = {
        'Random Forest': RandomForestClassifier(n_estimators=500,
                                                 random_state=SEED, n_jobs=-1),
        'SVM': SVC(kernel='rbf', C=10, gamma='scale', probability=True,
                    random_state=SEED),
        'DengueStackX': StackingClassifier(
            estimators=bl_drop,
            final_estimator=LogisticRegression(max_iter=2000, C=5.0,
                                                random_state=SEED),
            stack_method='predict_proba', cv=5, n_jobs=-1, passthrough=True),
    }
    for mname, m in test_models.items():
        m.fit(X_r_drop, y_res)
        yp  = m.predict(X_t_drop)
        acc = accuracy_score(y_test, yp)
        f1v = f1_score(y_test, yp, zero_division=0)
        sens_rows.append({'Removed Feature': feat, 'Model': mname,
                           'Accuracy': f"{acc*100:.2f}%",
                           'F1-Score': f"{f1v*100:.2f}%"})

sens_df = pd.DataFrame(sens_rows)
sens_df.to_csv(f'{OUT}/table06_sensitivity_analysis.csv', index=False)
sens_df
```
